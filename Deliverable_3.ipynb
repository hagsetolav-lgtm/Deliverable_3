Creating the character class.

```python
from character import Character

hero = Character("Jack")
hero.introduce()
print(hero.hp)
```

Task 1: Updating the character class.

```python
hero = Character("Jack", damage=10)
opponent = Character("Bob", hp_max=25)

hero.attack(opponent)
print(hero.hp, opponent.hp)

opponent.heal(50)
print(opponent.hp)

opponent.take_damage(100)
print(opponent.hp, opponent.is_alive)
print(hero.is_alive)
```

Updated the Character class so that the characters can take damage, attack and also heal, while staying between 0 and the maximum HP they can have. When a character attacks, it calls the take_damage() method. is_alive shows if the character is alive.

Task 2: Introducing randomness with a Dice class.

```python
from dice import Dice

d6 = Dice()
d20 = Dice(sides=20)

print(repr(d6))
print(repr(d20))
print("Six-sided dice roll", d6.roll())
print("Twenty-sided dice roll", d20())
```

Created a Dice class that defaults to 6 sides. `__repr__` is used to display the dice and `__call__` to roll by typing d20().

Task 3: Creating items.

```python
from items import Weapon

sword = Weapon("Sword", Dice(sides=6))
jack = Character("Jack", damage=10)
bob = Character("Bob", hp_max=25)

jack.attack(bob)
print("Bob's HP after the unarmed attack", bob.hp)

jack.weapon = sword
jack.attack(bob)
print("Bob's HP after the sword attack", bob.hp)
```

Created a Weapon class that uses a dice to calculate damage. Tested the attack before and after attaining a weapon. Unarmed attack deals 10 damage, while sword damage is between 1 and 6 depending on the dice roll.

Task 4: Character specialization.

```python
from character import Mage, Warrior

jack = Warrior("Jack")
staff = Weapon("Staff", Dice(sides=6))
bob = Mage("Bob", damage=10, weapon=staff, mana=10)

bob.attack(jack)
print("Jack's HP after the spell", jack.hp)
print("Bob's mana after the spell", bob.mana)

bob.attack(jack)
print("Jack's HP after the staff attack", jack.hp)
print("Bob's mana after the staff attack", bob.mana)
```

Created the Warrior class which reduces incoming damage. Created the warrior with the armor, then added function which applies the damage that gets through the armor.
Also created the Mage class which is able to cast spells using mana. Made a mage with a starting and maximum mana amount, and a function which casts a spell if enough mana is present.

Task 5: Running battles.

Unarmed battle first.

```python
from battle import Battle

jack = Warrior("Jack", hp_max=40)
bob = Mage("Bob", hp_max=30)
battle = Battle(hero=jack, enemy=bob)

battle.play_round()
print("After round 1:", jack.hp, bob.hp)

battle.simulate_battle()
print("Final HP:", jack.hp, bob.hp)
```

Battle with a weapon.

```python
sword = Weapon("Sword", Dice(sides=6))
jack = Warrior("Jack", hp_max=40, weapon=sword)
bob = Mage("Bob", hp_max=30)

battle = Battle(hero=jack, enemy=bob)
battle.simulate_battle()

winner = jack if jack.is_alive else bob
print("Winner:", winner.name)
print("Final HP:", jack.hp, bob.hp)
```

Created the class Battle, which simulates a battle between an enemy and a hero, which is turn-based. play_round lets the hero attack first, before the enemy can counterattack if alive. simulate_battle will simulate the battle between the character until one of them reach 0 HP.
